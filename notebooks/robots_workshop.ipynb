```python
from pathlib import Path

import pandas as pd

raw_data_dir = Path('./data/raw')
if not raw_data_dir.exists():
    raw_data_dir = Path('../data/raw')

# Load robot telemetry and shop data
try:
    electrical_current_df = pd.read_csv(raw_data_dir / 'electrical_current.csv')
    print('Shape of electrical_current:', electrical_current_df.shape)
except FileNotFoundError:
    print("Error: 'electrical_current.csv' not found.")
    electrical_current_df = pd.DataFrame()

try:
    robot_2_df = pd.read_csv(raw_data_dir / 'robot_2.csv')
    print('Shape of robot_2:', robot_2_df.shape)
except FileNotFoundError:
    print("Error: 'robot_2.csv' not found.")
    robot_2_df = pd.DataFrame()

try:
    robot_3_df = pd.read_csv(raw_data_dir / 'robot_3.csv')
    print('Shape of robot_3:', robot_3_df.shape)
except FileNotFoundError:
    print("Error: 'robot_3.csv' not found.")
    robot_3_df = pd.DataFrame()

try:
    robot_shop_df = pd.read_csv(raw_data_dir / 'robot_shop.csv')
    print('Shape of robot_shop:', robot_shop_df.shape)
except FileNotFoundError:
    print("Error: 'robot_shop.csv' not found.")
    robot_shop_df = pd.DataFrame()

# Combine robot samples and join shop details through the shop foreign key
robot_telemetry_df = pd.concat(
    [electrical_current_df, robot_2_df, robot_3_df],
    ignore_index=True,
)
robot_telemetry_shop_df = pd.merge(
    robot_telemetry_df,
    robot_shop_df,
    on='shop_id',
    how='left',
    validate='many_to_one',
)

print('\nFirst 5 rows after merging robot telemetry with shop data:')
print(robot_telemetry_shop_df.head())
print('Merged shape:', robot_telemetry_shop_df.shape)
```
